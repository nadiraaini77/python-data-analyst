##Isi Kolom Segment

```python
# Cari batas segment dari distribusi balance
q1 = df['Balance_Amount'].quantile(0.33)
q2 = df['Balance_Amount'].quantile(0.66)

# Function isi segment
def fill_segment(balance):

    if pd.isna(balance):
        return np.nan

    elif balance <= q1:
        return 'Mass'

    elif balance <= q2:
        return 'SME'

    else:
        return 'Priority'

# Isi hanya yang kosong
mask = df['Segment'].isna()

df.loc[mask, 'Segment'] = (
    df.loc[mask, 'Balance_Amount']
    .apply(fill_segment)
)
```

##Isi Kolom Dormant

```python
mask = df['Dormant_Flag'].isna()

df.loc[mask, 'Dormant_Flag'] = np.where(
    df.loc[mask, 'Txn_Count'] == 0,
    'Y',
    'N'
)
```

#Feature Engineering

##1. Tambah Tabel Target

###cek apakah ada target income yang 0

```python
target_achievement = (
    df['Interest_Income'].sum()
    /
    df['Target_Income'].sum()
) * 100

print(target_achievement)
```

```python
df['Target_Achievement'] = (
    (
        df['Interest_Income']
        /
        df['Target_Income']
    ) * 100
).round(2)
```

```python
df['Target_Achievement'].describe()
```

Seberapa besar pendapatan aktual yang berhasil dicapai dibanding target pendapatan yang sudah ditentukan bank.

Jadi dia dipakai buat ukur apakah performanya sudah sesuai ekspektasi atau belum.

#2. Tambah tabel profit

```python
df['Profit'] = (
    df['Interest_Income'] - df['Operational_Cost']
)
```

```python
df['Profit'].describe()
```

##3. Tambah Tabel Profit Margin

##cek apakah interest income ada yang 0

```python
(df['Interest_Income'] == 0).sum()
```

```python
df['Profit_Margin'] = (
    df['Profit']
    /
    df['Interest_Income']
) * 100
```

```python
df['Profit_Margin'].describe()
```

##4. Tambah Tabel Age Group

```python
bins = [18, 25, 40, 60, float('inf')]

labels = [
    'Gen Z',
    'Young Adult',
    'Mature',
    'Senior'
]

df['Age_Group'] = pd.cut(
    df['Customer_Age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```
